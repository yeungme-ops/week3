* Define a sub-DataFrame containing only those measurements that were taken at 28 days.

```python
df_sub = df[df["Age"] == 28]
df_sub
```

```python
df.query("Age == 28")
```

* Display a scatterplot in Matplotlib showing Cement vs Strength for this 28 days sample.

(I usually prefer different plotting libraries, but later I hope to add a regression line, and Matplotlib is convenient for this.)

```python
import matplotlib.pyplot as plt

plt.scatter(df_sub["Cement"], df_sub["Strength"])
plt.xlabel("Cement")
plt.ylabel("Strength")
plt.title("Concrete Strength at 28 Days")
plt.show()

fig, ax = plt.subplots()
ax.scatter(
    x = "Cement", 
    y = "Strength",
    data = df_sub
)
```

* Do you expect a fit linear regression coefficient to be positive or negative here?

## Introduction to scikit-learn

* Fit a regression line using scikit-learn.

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
type(reg)
```

```python
type(df_sub[["Cement"]])
```

```python
df_sub["Cement"].to_frame()
```

```python
reg.fit(df_sub[["Cement"]], df_sub[["Strength"]])
```

* Did the sign of the coefficient match your expectations?

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# reg2 = LinearRegression()
# reg2.coef
```

* How do we make predictions on new data using this model?

```python
arr = np.linspace(100, 500, 50)
```

```python
arr.shape
```

```python
reg.predict(arr.reshape((-1,1)))
```

```python
df_fake = pd.Series(arr, name = "Chris").to_frame()
```

```python
reg.predict(df_fake)
```

```python
df_fake = pd.Series(arr, name = "Cement").to_frame()
reg.predict(df_fake)
```

* Can you match those predictions using NumPy/Pandas directly?

```python
reg.coef_ * df_fake + reg.intercept_
```

* Plot the predicted line on the same axes using Matplotlib.

```python
afig, ax = plt.subplots()
ax.scatter(
    x = "Cement", 
    y = "Strength",
    data = df_sub,
)

ax.plot(
    arr,
    reg.coef_ * arr + reg.intercept_,
    color = "green"
)
```
